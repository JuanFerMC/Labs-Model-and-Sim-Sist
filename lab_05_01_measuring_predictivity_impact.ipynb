**your signal and target accuracy to achieve**

```python

s = ((np.random.normal(size=21)*5+.5).cumsum()+100).round(2)
acc = np.round(np.random.random()*.9+.1, 1)
print ("YOUR SIGNAL", s)
print ("THE ACCURACY YOUR SYNTHETIC PREDICTIONS MUST ACHIEVE: ", acc)
```

```python
perfect_prediction = (s[1:] > s[:-1]).astype(int)

# 3. Calculate how many values to flip to reach target accuracy
# Total items = 20. If acc=0.6, we need 12 correct, so we flip 8.
n_total = len(perfect_prediction)
n_flip = int(n_total * (1 - acc))

# 4. Randomly pick the indices to flip
flip_indices = np.random.permutation(n_total)[:n_flip]

# 5. Create the synthetic prediction by flipping the perfect one
# We use 1 - value to flip a 0 to 1 or a 1 to 0
prediction = perfect_prediction.copy()
prediction[flip_indices] = 1 - prediction[flip_indices]

# Verification
print(f"Synthetic Prediction: {prediction}")
```

```python
my_synthetic_prediction = perfect_prediction.copy()
my_synthetic_prediction[flip_indices] = 1 - my_synthetic_prediction[flip_indices]
actual_acc = np.sum(prediction == perfect_prediction) / n_total
print(f"Verified Accuracy: {actual_acc:.1f}")
print(f"Synthetic Prediction: {my_synthetic_prediction}")
```

```python
s = ((np.random.normal(size=21)*5+.5).cumsum()+100).round(2)
s
```

```python
import numpy as np

# Setup
s = np.array([102.98, 102.79, 106.53, 114.65, 113.98, 113.3, 121.7, 126.04, 124.19, 127.4,
              125.59, 123.76, 125.47, 116.4, 108.28, 105.96, 101.4, 103.47, 99.43, 92.87, 100.7])
perfect_p = (s[1:] > s[:-1]).astype(int)

pset_list = []
pnl_list = []

for i in range(1, 10):
    acc = i / 10.0
    n_wrong = 20 - int(20 * acc)

    # Generate prediction row with exact accuracy
    row = perfect_p.copy()
    flip_idx = np.random.permutation(20)[:n_wrong]
    row[flip_idx] = 1 - row[flip_idx]
    pset_list.append(row)

    # Strategy PNL Calculation
    diff = s[1:] - s[:-1]
    comm = s[:-1] * 0.01
    direction = (row - 0.5) * 2
    total_pnl = np.sum((direction * diff) - comm)
    pnl_list.append(round(total_pnl, 2))


```

```python
# a 9x20 numpy array
pset = np.array(pset_list)
print(pset)

# 9 elements numpy array or list
pnl = np.array(pnl_list)
print(pnl)
```
